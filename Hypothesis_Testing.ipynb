Question 1: What is a null hypothesis (H₀) and why is it important in hypothesis testing?

Sol - The null hypothesis (H₀) is a statement that there is no significant difference, effect, or relationship between the variables being studied.
- It is important in hypothesis testing because it provides a starting point for statistical analysis. We use sample data to determine whether there is enough evidence to reject H₀ or fail to reject H₀.
- Example: If we want to test whether a new teaching method improves student marks:
- H₀: The new teaching method does not improve student marks.
- H₁: The new teaching method improves student marks.
Thus, the null hypothesis helps us make an objective decision based on statistical evidence.

Question 2: What does the significance level (α) represent in hypothesis testing?

sol - The significance level (α) is the probability of rejecting the null hypothesis when it is actually true. It represents the acceptable risk of making a Type I error.
- Common significance levels are 0.05 (5%) and 0.01 (1%).
- For example, if α = 0.05, there is a 5% risk of rejecting a true null hypothesis.
- Therefore, the significance level helps determine whether the evidence from the sample is strong enough to reject the null hypothesis.



Question 3: Differentiate between Type I and Type II errors.

sol - Type I error occurs when we reject a true null hypothesis (H₀). It is also called a false positive.
- Example: Concluding that a new medicine works when it actually does not.
- Type II error occurs when we fail to reject a false null hypothesis (H₀). - It is also called a false negative.
- Example: Concluding that a new medicine does not work when it actually does.
- Type I Error
- Rejects a true H₀
- False positive
- Probability = α
- Type II Error
-	Fails to reject a false H₀
- 	False negative
-	Probability = β

In short: Type I error means detecting an effect that is not present, while Type II error means missing an effect that is actually present.

Question 4: Explain the difference between a one-tailed and two-tailed test. Give an example of each.

sol - A one-tailed test checks for a significant effect in one specific direction (either greater than or less than a certain value).
- Example: Testing whether a new teaching method increases the average student score.  
- H₀: μ ≤ 60
- H₁: μ > 60
- A two-tailed test checks whether there is a significant difference in either direction, without specifying whether it is higher or lower.
- Example: Testing whether the average student score is different from 60.  
- H₀: μ = 60
- H₁: μ ≠ 60
- In short: A one-tailed test looks in one direction, while a two-tailed test looks in both directions.

Question 5: A company claims that the average time to resolve a customer complaint is 10 minutes. A random sample of 9 complaints gives an average time of 12 minutes and a standard deviation of 3 minutes. At α = 0.05, test the claim.

sol - Given:
- Claimed population mean (μ) = 10 minutes
- Sample mean (x̄) = 12 minutes
- Sample standard deviation (s) = 3 minutes
- Sample size (n) = 9
- Significance level (α) = 0.05
- Since the sample size is small and the population standard deviation is unknown, we use a one-sample t-test.
- Step 1: State the hypotheses
- H₀: μ = 10 minutes
- H₁: μ ≠ 10 minutes
- This is a two-tailed test.
- Step 2: Calculate the t-value
- Formula:
- t = (x̄ − μ) / (s / √n)
- Substitute the values:
- t = (12 − 10) / (3 / √9)
- t = 2 / 1 = 2
- So, t = 2.00.
- Step 3: Compare with critical value
- Degrees of freedom:
- df = n − 1 = 9 − 1 = 8
- At α = 0.05 for a two-tailed test and df = 8:
- Critical t ≈ ±2.306
- Since:
|2.00| < 2.306
we fail to reject H₀.
- Conclusion
- There is not enough statistical evidence at the 5% significance level to reject the company’s claim that the average complaint resolution time is 10 minutes. Although the sample average is 12 minutes, the difference is not statistically significant.

Question 6: When should you use a Z-test instead of a t-test?

sol -
 Answer:
- A **Z-test** is used when the population standard deviation (σ) is known and the sample size is large (usually n ≥ 30).

- A **t-test** is used when the population standard deviation is unknown, especially when the sample size is small.

**In short:**
- **Z-test:** σ known and/or large sample
- **t-test:** σ unknown, especially small sample

Step 2: Calculate expected frequencies

Expected frequency = (Row total × Column total) / Grand total

- χ² = Σ (O − E)² / E

- χ² = 5 + 3.33 + 8.17 + 0.44
- χ² = 16.94
Step 4: Decision

Degrees of freedom:

- df = (2−1)(2−1) = 1
- Critical χ² value at α = 0.05 and df = 1 = 3.841
